```python
import murenn
import torch
from matplotlib import pyplot as plt
```

```python
udtcwt = murenn.UDTCWT(J=8)
x = torch.zeros(1, 1, 1000)
x[:, :, 500] = 1.0 
lp, bps = udtcwt(x)
```

```python
torch.sum(udtcwt.g0a)
```

```python
fig, axs = plt.subplots(4, 2, figsize=(10, 8))
for j, bp in enumerate(bps):
    print(bp.shape)
    axs[j // 2, j % 2].plot(bp.real.squeeze(), label='Real')
    axs[j // 2, j % 2].plot(bp.imag.squeeze(), label='Imaginary')
axs[0,0].legend()
```

```python

for bp in bps:
    psi_hat = torch.fft.fft(bp)
    plt.plot(torch.abs(psi_hat).squeeze().numpy()/max(torch.abs(psi_hat).squeeze().numpy()), label='Magnitude')
```
